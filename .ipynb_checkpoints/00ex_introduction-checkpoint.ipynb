1. The MickeyMouse problem

a) Write a program that prints the numbers from 1 to 100. But for multiples of 3 print `Mickey` instead of the corresponding number and for the multiples of 5 print `Mouse`. For numbers which are multiples of both three and five print `MickeyMouse`

b) Put the result in a tuple and substitute `Mickey` with `Donald` and `Mouse` with `Duck`

```python
"a)"
print("a)")
for i in range (1,101) :
    if i % 3 == 0 and i % 5 == 0 :
        print("MickeyMouse")
    elif i % 3 == 0 :
        print("Mickey")
    elif i % 5 == 0 :
        print("Mouse")
    else :
        print(i)

print()

"b)"
print("b)")
Disney = []
for i in range (1,101) :
    if i % 3 == 0 and i % 5 == 0 :
        Disney.append("MickeyMouse")
    elif i % 3 == 0 :
        Disney.append("Donald")
    elif i % 5 == 0 :
        Disney.append("Duck")
    else :
        Disney.append(i)
Disney = tuple(Disney)
print(Disney)
```

2\. The swap function

Write a function that swap the values of two input variables x and y (whatever the type). Try to do that also without a temporary variable

```python
def swap1 (x,y) :
    temp = x
    x = y
    y = temp
    return x, y

def swap2 (x,y) :
    x, y = y, x
    return x,y

x, y = swap2(2,3)
print(x,y)
```

3\. Computing the distance

Write a function that calculates and returns the euclidean distance between two points *u* and *v*, where *u* and *v* are both 2-tuples *(x,y)*. For example, if *u=(3,0)* and *v=(0,4)*, the function should return 5

```python
def Euclide_distance (u: tuple ,v: tuple) :
    if not isinstance(u, tuple):
        raise TypeError("L'argomento u deve essere una tupla")
    if not isinstance(v, tuple):
        raise TypeError("L'argomento v deve essere una tupla")

    distance = (abs(u[0]-v[0])**2+abs(u[1]-v[1])**2)**(1/2)
    return distance

u = (3,0)
v = (0,4)
d = Euclide_distance(u,v)
print(d)
```

4\. Counting letters

Write a program to calculate the number of times each character occurs in a given string *s*. Ignore differneces in capitalization

```python
s="Write a program that prints the numbers from 1 to 100. \
But for multiples of three print Mickey instead of the number and for the multiples of five print Mouse. \
For numbers which are multiples of both three and five print MickeyMouse"

characters = {}

for char in s :
    check = 0
    for i in characters :
        if char.casefold() == i :
            characters[i] = characters[i]+1
            check = 1
            break
    if check == 0 :
        characters[char.casefold()] = 1

for i in characters :
    print(i,characters[i])
```

5\. Isolating the unique

Write a function that determines and count the unique numbers in the list *l*

```python
l = [36, 45, 58, 3, 74, 96, 64, 45, 31, 10, 24, 19, 33, 86, 99, 18, 63, 70, 85,
 85, 63, 47, 56, 42, 70, 84, 88, 55, 20, 54, 8, 56, 51, 79, 81, 57, 37, 91,
 1, 84, 84, 36, 66, 9, 89, 50, 42, 91, 50, 95, 90, 98, 39, 16, 82, 31, 92, 41,
 45, 30, 66, 70, 34, 85, 94, 5, 3, 36, 72, 91, 84, 34, 87, 75, 53, 51, 20, 89, 51, 20]

def unique (lista) :
    numbers = {}
    uni_num = []
    for num in lista :
        check = 0
        for i in numbers :
            if num == i :
                numbers[i] = numbers[i]+1
                check = 1
                break
        if check == 0 :
            numbers[num] = 1
    for i in numbers :
        if numbers[i] == 1 :
            uni_num.append(i)
        
    return uni_num, len(uni_num)
    

print(unique(l))
```

6\. Combination of functions

Write two functions - one that returns the square of a number, and one that returns the cube. Now write a third function that returns the number raised to the 6th power using the two previous functions.

```python
def square (x) :
    return x*x

def cube (x) :
    return x*x*x

def sixth_pow (x) :
    return square(cube(x))

print(sixth_pow(2))
```

7\. Cubes

Create a list of the cubes of x for x in *[0, 10]* using:

a) a for loop

b) a list comprehension

```python
print("a)")
cubes = []
for i in range(11) :
    cubes.append(i**3)
print(cubes)

print("b)")
cubes = [i**3 for  i in range(11)]
print(cubes)
```

8\. Nested list comprehension

A Pythagorean triple is an integer solution to the Pythagorean theorem $a^2+b^2=c^2$. The first Pythagorean triple is (3,4,5). Find and put in a tuple all unique Pythagorean triples for the positive integers a, b and c less than 100.

```python
pyt = []
for a in range(1,100) :
    for b in range(a,100) :
        c = (a**2+b**2)**(1/2)
        if c<100 and c-int(c) == 0 :
            pyt.append((a,b,int(c)))
pyt = tuple(pyt)
print(pyt)
```

9\. Normalization

Write a function that takes a tuple of numbers and returns it with the entries normalized to one

```python
def normalization (t) :
    t = list(t)
    s = 0
    for i in range(len(t)) :
        s += t[i]**2
    for i in range(len(t)) :
        t[i] /= (s**0.5)
    return t

tupla = (1,2,3,4)
print(normalization(tupla))
```
