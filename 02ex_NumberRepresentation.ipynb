1\. Write a function that converts number representation, bin<->dec<->hex. (Clearly using the corresponding python built-in functions is not fair..)

```python
n = 67596759
decimal_n, binary_n, hexadecimal_n = int(n), bin(n), hex(n)

def bin_to_dec (bin_n) :
    d = 0
    for i in range(2, len(bin_n)) :
        if bin_n[i] == '1' :
            d += 2**(len(bin_n)-1-i)
    if bin_n[0] == '-' :
        d = -d
    return d


def dec_to_bin (int_n) :
    b = ''
    if int_n == 0 :
        return '0b0'
    elif int_n < 0 :
        int_n *= -1
        b = '-'
    i, r = 0, 0  # power index and carry
    # decomposition in powers of 2
    while int_n != 1 :
        if int_n % 2 == 0 :
            int_n = int_n // 2
            i += 1
        else :
            int_n -= 1
            r += 2**i
    # setting array lenght and first digits
    p = ['0'] * (i+3)
    p[0] = '0'
    p[1] = 'b'
    p[2] = '1'
    # decomposition of the carry and further digits
    R = r
    while R > 0 :
        i, r = 0, 0
        while R != 1 :
            if R % 2 == 0 and R != 0:
                R = R // 2
                i += 1
            else :
                R -= 1
                r += 2**i
        p[-1-i] = '1'
        R = r
    # converting to string
    for j in p :
        b += j
    return b


def hex_to_dec (hex_n) :
    h = 0
    for i in range(2, len(hex_n)) :
        if hex_n[i] == '1' : h += 16**(len(hex_n)-1-i)
        elif hex_n[i] == '2' : h += 2*16**(len(hex_n)-1-i)
        elif hex_n[i] == '3' : h += 3*16**(len(hex_n)-1-i)
        elif hex_n[i] == '4' : h += 4*16**(len(hex_n)-1-i)
        elif hex_n[i] == '5' : h += 5*16**(len(hex_n)-1-i)
        elif hex_n[i] == '6' : h += 6*16**(len(hex_n)-1-i)
        elif hex_n[i] == '7' : h += 7*16**(len(hex_n)-1-i)
        elif hex_n[i] == '8' : h += 8*16**(len(hex_n)-1-i)
        elif hex_n[i] == '9' : h += 9*16**(len(hex_n)-1-i)
        elif hex_n[i] == 'a' : h += 10*16**(len(hex_n)-1-i)
        elif hex_n[i] == 'b' : h += 11*16**(len(hex_n)-1-i)
        elif hex_n[i] == 'c' : h += 12*16**(len(hex_n)-1-i)
        elif hex_n[i] == 'd' : h += 13*16**(len(hex_n)-1-i)
        elif hex_n[i] == 'e' : h += 14*16**(len(hex_n)-1-i)
        elif hex_n[i] == 'f' : h += 15*16**(len(hex_n)-1-i)
    if hex_n[0] == '-' :
        h = -h
    return h


def dec_to_hex (dec_n) :
    x = ''
    if dec_n == 0:
        return '0x0'
    while dec_n != 0 :
        r = dec_n % 16
        if r < 10 : r = str(r)
        elif r == 10 : r = 'a'
        elif r == 11 : r = 'b'
        elif r == 12 : r = 'c'
        elif r == 13 : r = 'd'
        elif r == 14 : r = 'e'
        elif r == 15 : r = 'f'
        x = r + x
        dec_n = dec_n // 16
    x = '0x' + x
    return x
    


print("Decimal representation of", binary_n, ":", bin_to_dec(binary_n))
print("Binary representation of", decimal_n, ":", dec_to_bin(decimal_n))
print("Decimal representation of", hexadecimal_n, ":", hex_to_dec(hexadecimal_n))
print("Hexadecimal representation of", decimal_n, ":", dec_to_hex(decimal_n))
```

2\. Write a function that converts a 32 bit word into a single precision floating point (i.e. interprets the various bits as sign, mantissa and exponent)

```python
word = '01000001001011000000000000000000'

def float_converter (bits) :
    # mantissa
    f = 0
    for i in range(9,32) :
        if bits[i] == '1' :
            f += 2**(8-i)
    # exponent
    e = 0
    for i in range(1,9) :
        if bits[i] == '1' :
            e += 2**(8-i)
    # sign
    if bits[0] == '1' :
        s = 1
    if bits[0] == '0' :
        s = 0
    # floating point number
    x = (-1)**s * (1+f) * 2**(e-127)
    return x

print(float_converter(word))
```

3\. Write a program to determine the underflow and overflow limits (within a factor of 2) for python on your computer. 

**Tips**: define two variables inizialized to 1 and halve/double them enough time to exceed the under/over-flow limits  

```python
# underflow
u = 1.0
while True :
    n = u / 2
    if n == 0 :
        break
    u = n
    
print("Underflow limit:", u)

# overflow
o = 1.0
while True:
    m = o * 2
    if m == float('inf'):
        break
    o = m

print("Overflow limit:", o)
```

5\. Write a function that takes in input three parameters $a$, $b$ and $c$ and prints out the two solutions to the quadratic equation $ax^2+bx+c=0$ using the standard formula:
$$
x=\frac{-b\pm\sqrt{b^2-4ac}}{2a}
$$

(a) use the program to compute the solution for $a=0.001$, $b=1000$ and $c=0.001$

(b) re-express the standard solution formula by multiplying top and bottom by $-b\mp\sqrt{b^2-4ac}$ and again find the solution for $a=0.001$, $b=1000$ and $c=0.001$. How does it compare with what previously obtained? Why?

(c) write a function that compute the roots of a quadratic equation accurately in all cases

```python
def quadratic (a,b,c) :
    x1 = (-b + (b**2 - 4*a*c)**(1/2)) / (2*a)
    x2 = (-b - (b**2 - 4*a*c)**(1/2)) / (2*a)
    return x1, x2

def quadratic2 (a,b,c) :
    x1 = (2*c) / (-b - (b**2 - 4*a*c)**(1/2))
    x2 = (2*c) / (-b + (b**2 - 4*a*c)**(1/2))
    return x1, x2

def quadratic3 (a,b,c) :
    x1 = (2*c) / (-b - (b**2 - 4*a*c)**(1/2))
    x2 = (-b - (b**2 - 4*a*c)**(1/2)) / (2*a)
    return x1, x2

print("a)")
print(quadratic(0.001, 1000, 0.001))
print("b)")
print(quadratic2(0.001, 1000, 0.001))
print("c)")
print(quadratic3(0.001, 1000, 0.001))
```

6\. Write a program that implements the function $f(x)=x(x−1)$

(a) Calculate the derivative of the function at the point $x = 1$ using the derivative definition:

$$
\frac{{\rm d}f}{{\rm d}x} = \lim_{\delta\to0} \frac{f(x+\delta)-f(x)}{\delta}
$$

with $\delta = 10^{−2}$. Calculate the true value of the same derivative analytically and compare with the answer your program gives. The two will not agree perfectly. Why not?

(b) Repeat the calculation for $\delta = 10^{−4}, 10^{−6}, 10^{−8}, 10^{−10}, 10^{−12}$ and $10^{−14}$. How does the accuracy scales with $\delta$?

```python
def f (x) :
    return x*(x-1)

def f1 (x) :
    return 2*x-1

print("a)")
d = 10**-2
x = 1
df = (f(x+d)-f(x))/d
Df = f1(x)
print("f'(1) using derivative definition:",df)
print("f'(1) using analytical definition:", Df)

print("b)")
for i in range(4,15,2) :
    d = 10**-i
    df = (f(x+d)-f(x))/d
    print(f"delta:{d:.1e};", "f'(1) =", df)
```

7\. Consider the integral of the semicircle of radius 1:
$$
I=\int_{-1}^{1} \sqrt(1-x^2) {\rm d}x
$$
which it's known to be $I=\frac{\pi}{2}=1.57079632679...$.
Alternatively we can use the Riemann definition of the integral:
$$
I=\lim_{N\to\infty} \sum_{k=1}^{N} h y_k 
$$

with $h=2/N$ the width of each of the $N$ slices the domain is divided into and where
$y_k$ is the value of the function at the $k-$th slice.

(a) Write a programe to compute the integral with $N=100$. How does the result compares to the true value?

(b) How much can $N$ be increased if the computation needs to be run in less than a second? What is the gain in running it for 1 minute? 


```python
import numpy as np
import time

def f (x) :
    return (1-x**2)**(1/2)

N = 100
h = 2/N

x = np.linspace(-1,1,N)
y = f(x)
I = sum(y*h)

print("Integral using Riemann definition:   ", I)
print("Integral using analytical definition:", np.pi/2)
```
